# Stage 7 – XGBoost Hyperparameter Optimization

This notebook optimizes the baseline XGBoost model developed in
Stage 6.

RandomizedSearchCV is used to explore different combinations of
XGBoost hyperparameters using stratified 5-fold cross-validation.

The tuned model is then compared with the Stage 6 baseline and
evaluated on the held-out test set.

```python
from pathlib import Path
import sys

current = Path.cwd()

while current != current.parent:
    if (current / "src").is_dir():
        PROJECT_ROOT = current
        break
    current = current.parent
else:
    raise RuntimeError(
        "Could not find project root containing 'src'."
    )

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
import pandas as pd
import joblib

from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    recall_score,
    precision_score,
    f1_score,
    confusion_matrix
)

from src.config import RANDOM_STATE, TARGET
from src.data.data_cleaning import clean_and_split
from src.models.xgboost_model import build_xgboost
from src.preprocessing.feature_engineering import build_pipeline
from src.evaluation.metrics import evaluate_model, results_to_row

import xgboost

print("All imports successful.")
print("XGBoost version:", xgboost.__version__)
```

```python
train_df, test_df = clean_and_split()

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
```

```python
X_train = train_df.drop(columns=[TARGET])
y_train = train_df[TARGET]

X_test = test_df.drop(columns=[TARGET])
y_test = test_df[TARGET]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

## 3. Build the XGBoost Pipeline for Tuning

A fresh XGBoost model is created and placed inside the same
preprocessing and feature-engineering pipeline used in Stage 6.

```python
xgb_tuning = build_xgboost()

xgb_tuning_pipeline = build_pipeline(
    xgb_tuning
)

print(xgb_tuning_pipeline)
```

## 4. Define the Hyperparameter Search Space

The following hyperparameters are explored:

- Number of boosting trees
- Learning rate
- Maximum tree depth
- Minimum child weight
- Row subsampling
- Feature subsampling
- Minimum loss reduction (`gamma`)
- L1 regularization (`reg_alpha`)
- L2 regularization (`reg_lambda`)

```python
param_distributions = {
    "model__n_estimators": [
        200,
        300,
        400,
        500,
        700
    ],

    "model__learning_rate": [
        0.01,
        0.03,
        0.05,
        0.08,
        0.10
    ],

    "model__max_depth": [
        3,
        4,
        5,
        6,
        8
    ],

    "model__min_child_weight": [
        1,
        3,
        5,
        10
    ],

    "model__subsample": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    "model__colsample_bytree": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    "model__gamma": [
        0,
        0.1,
        0.3,
        0.5
    ],

    "model__reg_alpha": [
        0,
        0.01,
        0.1,
        1
    ],

    "model__reg_lambda": [
        1,
        2,
        5,
        10
    ]
}
```

## 5. Configure the Hyperparameter Search

RandomizedSearchCV is used instead of testing every possible
combination.

Thirty randomly selected parameter combinations are evaluated
using 5-fold cross-validation.

ROC-AUC is used as the optimization metric.

```python
xgb_tuning_pipeline.set_params(
    model__n_jobs=1
)
```

```python
random_search = RandomizedSearchCV(
    estimator=xgb_tuning_pipeline,
    param_distributions=param_distributions,
    n_iter=30,
    scoring="roc_auc",
    cv=5,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    verbose=2,
    return_train_score=True
)
```

## 6. Run Hyperparameter Search

Each sampled hyperparameter configuration is evaluated using
5-fold cross-validation on the training data.

The held-out test set is not used during this process.

```python
random_search.fit(
    X_train,
    y_train
)
```

## 7. Identify the Best Hyperparameter Configuration

```python
print(
    "Best CV ROC-AUC:",
    random_search.best_score_
)
```

```python
print("Best XGBoost parameters:\n")

for parameter, value in random_search.best_params_.items():
    print(f"{parameter}: {value}")
```


## 8. Obtain the Tuned XGBoost Pipeline

```python
xgb_tuned_pipeline = random_search.best_estimator_

print(xgb_tuned_pipeline)
```

## 9. Evaluate the Tuned XGBoost Model

The tuned model is evaluated using the same evaluation function
used for the Stage 6 baseline.

This ensures that baseline and tuned results are directly comparable.

```python
tuned_results = evaluate_model(
    xgb_tuned_pipeline,
    X_train,
    y_train,
    model_name="XGBoost Tuned"
)

tuned_results
```

## 10. Compare XGBoost Baseline and Tuned Model

```python
experiments_path = (
    PROJECT_ROOT
    / "experiments"
    / "results"
    / "experiments.csv"
)

all_results = pd.read_csv(experiments_path)

xgb_baseline_row = all_results[
    all_results["model_name"] == "XGBoost Baseline"
].copy()

tuned_row = results_to_row(
    tuned_results
)

xgb_comparison = pd.concat(
    [
        xgb_baseline_row,
        tuned_row
    ],
    ignore_index=True
)

xgb_comparison
```

## 11. Save XGBoost Baseline vs Tuned Results

```python
results_dir = (
    PROJECT_ROOT
    / "experiments"
    / "results"
)

results_dir.mkdir(
    parents=True,
    exist_ok=True
)

xgb_comparison.to_csv(
    results_dir / "xgboost_baseline_vs_tuned.csv",
    index=False
)

print(
    "Saved:",
    results_dir / "xgboost_baseline_vs_tuned.csv"
)
```

## 12. Save All Hyperparameter Search Results

All tested configurations are saved so that the optimization
experiment can be reproduced and analysed later.

```python
cv_results = pd.DataFrame(
    random_search.cv_results_
)

cv_results.to_csv(
    results_dir / "xgboost_random_search.csv",
    index=False
)

print(
    "Saved:",
    results_dir / "xgboost_random_search.csv"
)
```

## 13. Save the Best Hyperparameters

```python
best_params_df = pd.DataFrame(
    [
        {
            "parameter": parameter,
            "value": value
        }
        for parameter, value
        in random_search.best_params_.items()
    ]
)

best_params_df.to_csv(
    results_dir / "xgboost_best_parameters.csv",
    index=False
)

best_params_df
```

## 14. Fit the Selected XGBoost Pipeline on the Complete Training Set

The best hyperparameter configuration identified through
cross-validation is fitted on the complete training dataset.

The test dataset remains untouched until final evaluation.

```python
xgb_tuned_pipeline.fit(
    X_train,
    y_train
)
```

## 15. Evaluate the Tuned XGBoost Model on the Held-Out Test Set

```python
y_test_proba = (
    xgb_tuned_pipeline
    .predict_proba(X_test)[:, 1]
)

y_test_pred = (
    y_test_proba >= 0.5
).astype(int)
```

```python
final_roc_auc = roc_auc_score(
    y_test,
    y_test_proba
)

final_pr_auc = average_precision_score(
    y_test,
    y_test_proba
)

final_recall = recall_score(
    y_test,
    y_test_pred,
    zero_division=0
)

final_precision = precision_score(
    y_test,
    y_test_pred,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    y_test_pred,
    zero_division=0
)

final_cm = confusion_matrix(
    y_test,
    y_test_pred
)

print("Final XGBoost Test Results")
print("=" * 40)
print(f"ROC-AUC   : {final_roc_auc:.4f}")
print(f"PR-AUC    : {final_pr_auc:.4f}")
print(f"Recall    : {final_recall:.4f}")
print(f"Precision : {final_precision:.4f}")
print(f"F1 Score  : {final_f1:.4f}")

print("\nConfusion Matrix:")
print(final_cm)
```

## 16. Save the Tuned XGBoost Model

The complete fitted pipeline is saved so that the exact preprocessing,
feature engineering and XGBoost model can later be loaded by the
prediction system.

```python
model_dir = (
    PROJECT_ROOT
    / "models"
    / "tuned"
)

model_dir.mkdir(
    parents=True,
    exist_ok=True
)

model_path = (
    model_dir
    / "xgboost_tuned.joblib"
)

joblib.dump(
    xgb_tuned_pipeline,
    model_path
)

print("Saved:", model_path)
```

## 17. Verify the Saved Model

The saved model is loaded again and used to generate predictions
for a few test observations.

This confirms that the saved pipeline can be restored successfully.

```python
loaded_model = joblib.load(
    model_path
)

test_probability = (
    loaded_model
    .predict_proba(X_test.head(5))[:, 1]
)

print(test_probability)
```
