### What is "Data Understanding"?
In data mining (CRISP-DM methodology), **Data Understanding** is the foundational phase where we inspect raw data to:
1. Understand the size and structure of the dataset.
2. Check the data types of individual features.
3. Identify missing or corrupted data.
4. Detect data quality anomalies (e.g., numbers stored as text strings).

## 2. Loading the Dataset
We load the raw CSV file using `pd.read_csv()`.

> **Why `low_memory=False`?**  
> Pandas reads large CSV files in chunks and tries to guess the data type of each column based on early rows. If a column contains mostly numbers in early rows but has symbols (such as `$` or `x`) in later rows, Pandas will issue a `DtypeWarning`. Specifying `low_memory=False` instructs Pandas to load the entire column into memory before determining data types, preventing incorrect parsing and type warnings.


```python
# Relative path handling: works whether launched from the workspace root or inside the notebooks folder
data_path = 'data/raw/Train_Dataset.csv' if os.path.exists('data/raw/Train_Dataset.csv') else '../data/raw/Train_Dataset.csv'

# Load raw dataset
df = pd.read_csv(data_path, low_memory=False)

print(f"Dataset loaded successfully from: '{data_path}'")
```

## 4. Missing Values Analysis
In real-world data collection, applicants may skip optional questions, records might be lost, or systems might fail to capture certain fields. In Pandas, missing data is represented as `NaN` (Not a Number) or `None`.

Here, we:
1. Count the number of missing values per column with `df.isnull().sum()`.
2. Compute the percentage of missing values: `(missing_count / total_rows) * 100`.
3. Display all columns containing missing values, sorted from highest to lowest percentage.


```python
# Count missing values per column
missing_count = df.isnull().sum()

# Calculate percentage of missing values
missing_pct = (missing_count / len(df)) * 100

# Combine into a clean summary DataFrame
missing_summary = pd.DataFrame({
    'Column_Name': df.columns,
    'Missing_Count': missing_count.values,
    'Missing_Percentage (%)': missing_pct.values.round(2)
})

# Filter only columns that have at least one missing value and sort descending
missing_summary_filtered = (
    missing_summary[missing_summary['Missing_Count'] > 0]
    .sort_values(by='Missing_Percentage (%)', ascending=False)
    .reset_index(drop=True)
)

print(f"Total columns with missing values: {len(missing_summary_filtered)} out of {num_cols}")
missing_summary_filtered
```

### 💡 Plain English Interpretation of Missing Values:
1. **High Missingness (>50%):**
   - `Own_House_Age` (~66% missing): This column has high missingness primarily because not all applicants own a house (indicated by `House_Own`). If someone rents or lives with parents, their home age is naturally absent.
   - `Score_Source_1` (~56% missing): An external credit bureau score. Many applicants simply do not have an available score from this specific agency.
2. **Moderate Missingness (10% - 50%):**
   - Features like `Social_Circle_Default`, `Client_Occupation`, and `Score_Source_3` have missing rates between 18% and 53%.
3. **Low Missingness (<5%):**
   - Features like `Credit_Bureau`, `Phone_Change`, `Loan_Annuity`, and `Credit_Amount` have very few missing records (<5%).
4. **Action for Preprocessing:**
   - In `03_data_preprocessing.ipynb`, we will apply appropriate strategies: imputation (mean, median, mode) for moderate/low missingness, indicator flags, or domain-specific handling.


## 5. Identifying Numerical Columns Stored Incorrectly as Object / String

In tabular data mining, numerical features (such as income, loan amount, age, and scores) must be stored as numbers (`int64` or `float64`) so that:
- We can perform mathematical operations (mean, median, standard deviation).
- Machine learning algorithms can compute distances and weights.

However, if even a **single row** in a column contains a non-numeric character (such as `'$'`, `'#'`, `'@'`, `'x'`, or `'#VALUE!'`), Pandas will parse the **entire column** as `object` or `string`!

Let's find out which columns are affected and examine what dirty characters are causing this issue.


```python
# Expected numerical columns that should be int or float based on domain meaning
candidate_numerical_cols = [
    'Client_Income',
    'Credit_Amount',
    'Loan_Annuity',
    'Population_Region_Relative',
    'Age_Days',
    'Employed_Days',
    'Registration_Days',
    'ID_Days',
    'Score_Source_3'
]

# Check their current data types in the loaded DataFrame
dirty_columns_analysis = []

for col in candidate_numerical_cols:
    current_dtype = str(df[col].dtype)
    
    # Try converting to numeric using pd.to_numeric with errors='coerce'
    # Any value that cannot be converted to a number becomes NaN
    coerced_series = pd.to_numeric(df[col], errors='coerce')
    
    # Identify entries where the original was NOT null, but conversion resulted in NaN
    corrupted_mask = df[col].notna() & coerced_series.isna()
    corrupted_count = corrupted_mask.sum()
    unique_corrupted_values = df.loc[corrupted_mask, col].unique().tolist()
    
    dirty_columns_analysis.append({
        'Feature': col,
        'Current_Dtype': current_dtype,
        'Expected_Dtype': 'float64 / int64',
        'Corrupted_Rows_Count': corrupted_count,
        'Dirty_Characters_Found': unique_corrupted_values
    })

dirty_cols_df = pd.DataFrame(dirty_columns_analysis)
dirty_cols_df
```

### 🔎 Inspecting Sample Corrupted Rows
Let's see some concrete examples of rows containing these invalid characters to understand what happened during data entry or CSV export.


```python
from IPython.display import display

# Show sample rows where Client_Income has non-numeric characters
print("--- Sample rows with corrupted 'Client_Income' ---")
display(df[df['Client_Income'] == '$'][['ID', 'Client_Income', 'Client_Income_Type', 'Default']].head(3))

# Show sample rows where Age_Days or Employed_Days has non-numeric characters
print("\n--- Sample rows with corrupted 'Age_Days' or 'Employed_Days' ---")
display(df[(df['Age_Days'] == 'x') | (df['Employed_Days'] == 'x')][['ID', 'Age_Days', 'Employed_Days', 'Default']].head(3))

# Show sample rows where Score_Source_3 has non-numeric characters
print("\n--- Sample rows with corrupted 'Score_Source_3' ---")
display(df[df['Score_Source_3'] == '&'][['ID', 'Score_Source_3', 'Default']].head(3))
```

### 💡 Plain English Explanation for Undergraduates:
Why did these columns get classified as `object`?
- **`Client_Income` & `Credit_Amount`:** Contain `$` characters instead of pure numbers. When someone entered `$` or exported from Excel with raw currency formatting, Pandas couldn't treat it as a number.
- **`Loan_Annuity`:** Contains `$` symbols and `'#VALUE!'` (a classic formula error from Microsoft Excel).
- **`Age_Days`, `Employed_Days`, `Registration_Days`, `ID_Days`:** Contain `'x'` characters, likely used by data entry operators as a placeholder for "unknown" or "unverified".
- **`Population_Region_Relative`:** Contains `'@'` and `'#'` special characters.
- **`Score_Source_3`:** Contains `'&'` characters.

**Key Learning Point:**  
Before feeding data into machine learning models, these invalid characters must be cleaned or replaced with `NaN`, and the columns must be converted to float/integer (`pd.to_numeric()`). This will be performed in our Data Preprocessing stage (`03_data_preprocessing.ipynb`).


## 6. Target Variable Inspection: `Default`
In this project, our goal is to predict whether a customer will default on their vehicle loan.
- **`Default = 1`:** The client defaulted on payments (high risk).
- **`Default = 0`:** The client repaid on time (low risk).

Let's check the distribution of the target class.


```python
# Target variable class distribution
default_counts = df['Default'].value_counts()
default_percentages = df['Default'].value_counts(normalize=True) * 100

target_df = pd.DataFrame({
    'Count': default_counts,
    'Percentage (%)': default_percentages.round(2)
})
target_df.index = ['Non-Default (0)', 'Default (1)']
target_df
```

### 💡 Target Variable Takeaway:
- Over 91% of applicants did not default (`0`), while approximately 8% defaulted (`1`).
- This means our dataset is **imbalanced**. Class imbalance is typical in credit and fraud detection, and we will address it using appropriate evaluation metrics (ROC-AUC, Precision, Recall, F1-Score) and modeling techniques in upcoming notebooks.
